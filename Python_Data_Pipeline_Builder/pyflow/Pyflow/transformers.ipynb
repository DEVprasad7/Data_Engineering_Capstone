```python
# Use defaultdict for hourly fare analysis with pythonn defaultdict. 
from collections import defaultdict
import math
import statistics

cleaned_set = subset.query('fare_amount >= 0')

# 1. Group fares by integer hour duration (0 = <1 hr, 1 = 1-2 hrs, etc.)
duration_fares = defaultdict(list)

for duration_hr, fare in zip(cleaned_set['Trip_time_hr'], cleaned_set['fare_amount']):
    # math.floor() handles negative durations safely if any exist
    hour_bucket = math.floor(duration_hr)
    duration_fares[hour_bucket].append(fare)

# 2. Calculate metrics for each duration bucket
summary = []
for bucket in sorted(duration_fares):
    fares = duration_fares[bucket]
    summary.append({
        'duration_range': f"{bucket} to {bucket + 1} hrs",
        'trip_count': len(fares),
        'avg_fare': round(statistics.mean(fares), 2),
        'median_fare': round(statistics.median(fares), 2),
        'min_fare': min(fares),
        'max_fare': max(fares)
    })

duration_summary_cleaned_set = pd.DataFrame(summary)
print(duration_summary_cleaned_set.to_string(index=True))
# Use defaultdict for hourly fare analysis with pythonn defaultdict. 
```

```python
# Use deque as a rolling buffer to calculate 1d rolling revenue. 
from collections import deque
from datetime import timedelta

subset1 = df[['tpep_pickup_datetime', 'total_amount']]
subset1 = subset1.query('total_amount > 0')
subset1 = subset1.sort_values('tpep_pickup_datetime').reset_index(drop=True)

window = deque()
running_total = 0
time_limit = timedelta(days=1)
rev_list = []


for ts, amt in zip(subset1["tpep_pickup_datetime"], subset1["total_amount"]):
    window.append((ts, amt))
    running_total += amt

    cutoff = ts - time_limit
    while window and window[0][0] <= cutoff:
        _, old_amt = window.popleft()
        running_total -= old_amt

    rev_list.append(running_total)

subset1["rolling_1d_revenue"] = rev_list
subset1
```

```python
# Design TripRecord class for deduplication.
from dataclasses import dataclass
from typing import Any
import pandas as pd


@dataclass(frozen=True, slots=True)
class TripRecord:
    VendorID: int
    tpep_pickup_datetime: str
    tpep_dropoff_datetime: str
    PULocationID: int
    DOLocationID: int
    trip_distance: float
    total_amount: float

    @classmethod
    def from_dict(cls, row: dict[str, Any]) -> "TripRecord":
        return cls(
            VendorID=int(row["VendorID"]),
            tpep_pickup_datetime=str(row["tpep_pickup_datetime"]),
            tpep_dropoff_datetime=str(row["tpep_dropoff_datetime"]),
            PULocationID=int(row["PULocationID"]),
            DOLocationID=int(row["DOLocationID"]),
            trip_distance=round(float(row["trip_distance"]), 2),
            total_amount=round(float(row["total_amount"]), 2),
        )


# Deduplicate using a set
records = (TripRecord.from_dict(row) for row in df.to_dict(orient="records"))
unique_records = set(records)

# Rebuild DataFrame with just the core fields
df_clean = pd.DataFrame([
    {
        "VendorID": r.VendorID,
        "tpep_pickup_datetime": r.tpep_pickup_datetime,
        "tpep_dropoff_datetime": r.tpep_dropoff_datetime,
        "PULocationID": r.PULocationID,
        "DOLocationID": r.DOLocationID,
        "trip_distance": r.trip_distance,
        "total_amount": r.total_amount,
    }
    for r in unique_records
])
```
